```python
import numpy as np, matplotlib.pyplot as plt
rng=np.random.default_rng(7); X=rng.random((100,2)); y=(X[:,0]+X[:,1]>.9).astype(int)
w=np.zeros(2); b=0
for epoch in range(20):
 for x,t in zip(X,y): pred=int(w@x+b>=0); w+=.1*(t-pred)*x; b+=.1*(t-pred)
print('Weights/bias:',w,b)
xx=np.linspace(0,1,100); yy=-(w[0]*xx+b)/w[1]
plt.scatter(X[:,0],X[:,1],c=y); plt.plot(xx,yy); plt.show()
Xxor=np.array([[0,0],[0,1],[1,0],[1,1]]); yxor=np.array([0,1,1,0]); w=np.zeros(2); b=0
for _ in range(20):
 for x,t in zip(Xxor,yxor): p=int(w@x+b>=0); w+=.1*(t-p)*x; b+=.1*(t-p)
print('XOR predictions:',(Xxor@w+b>=0).astype(int),'target:',yxor)
for z in [-2,0,2]: print(z,'step',int(z>=0),'sigmoid',1/(1+np.exp(-z)),'ReLU',max(0,z))
```
