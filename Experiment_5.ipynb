```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
rng=np.random.default_rng(5); X=np.linspace(0,1,150)[:,None]; y=np.sin(2*np.pi*X[:,0])+rng.normal(0,.25,150)
Xt,Xv,yt,yv=train_test_split(X,y,test_size=.3,random_state=1)
degrees=[1,3,9,15]; tr=[]; va=[]
plt.scatter(X,y,s=12)
for d in degrees:
 m=make_pipeline(PolynomialFeatures(d),LinearRegression()).fit(Xt,yt); tr.append(np.mean((m.predict(Xt)-yt)**2)); va.append(np.mean((m.predict(Xv)-yv)**2)); plt.plot(X,m.predict(X),label=str(d))
plt.legend(title='Degree'); plt.show()
plt.plot(degrees,tr,'o-',label='Train'); plt.plot(degrees,va,'o-',label='Validation'); plt.legend(); plt.show()
print('Errors:',list(zip(degrees,tr,va)),'Best degree:',degrees[np.argmin(va)])
```
