```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
rng=np.random.default_rng(2); X=rng.normal(size=(200,5))*[1,2,10,20,3]+[13,4,100,180,25]
df=pd.DataFrame(X,columns=['Hb','RBC','Glucose','Cholesterol','BMI'])
Z=(df-df.mean())/df.std()
cov=Z.cov(); corr=Z.corr(); print(cov); print(corr)
for M,title in [(cov,'Covariance'),(corr,'Correlation')]:
 plt.imshow(M,cmap='coolwarm'); plt.colorbar(); plt.xticks(range(5),df.columns,rotation=45); plt.yticks(range(5),df.columns); plt.title(title); plt.tight_layout(); plt.show()
w,v=np.linalg.eigh(cov); order=np.argsort(w)[::-1]; w=w[order]; v=v[:,order]
print('Eigenvalues:',w,'\nVariance %:',w/w.sum()*100)
plt.scatter(Z.iloc[:,0],Z.iloc[:,1],alpha=.4); origin=np.zeros(2)
for k in range(2): plt.arrow(0,0,v[0,k],v[1,k],width=.02);
plt.title('Data and leading eigenvectors'); plt.show()
```
