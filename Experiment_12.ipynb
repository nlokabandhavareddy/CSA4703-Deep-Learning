```python
import tensorflow as tf, numpy as np, time
(x,y),(xt,yt)=tf.keras.datasets.cifar10.load_data(); x=x[:8000].astype('float32')/255; y=y[:8000].ravel(); xt=xt[:2000].astype('float32')/255; yt=yt[:2000].ravel()
results=[]
for filters,k,pool in [(32,3,'max'),(32,5,'avg'),(64,3,'max')]:
 m=tf.keras.Sequential([tf.keras.layers.Input((32,32,3)),tf.keras.layers.Conv2D(filters,k,activation='relu'),tf.keras.layers.BatchNormalization(),tf.keras.layers.MaxPooling2D((2,2)) if pool=='max' else tf.keras.layers.AveragePooling2D((2,2)),tf.keras.layers.Conv2D(filters*2,3,activation='relu'),tf.keras.layers.Flatten(),tf.keras.layers.Dense(64,activation='relu'),tf.keras.layers.Dense(10,activation='softmax')])
 m.compile('adam','sparse_categorical_crossentropy',metrics=['accuracy']); t=time.time(); m.fit(x,y,epochs=2,batch_size=64,verbose=0); dur=time.time()-t; acc=m.evaluate(xt,yt,verbose=0)[1]; results.append((filters,k,pool,acc,m.count_params(),dur))
print('filters,kernel,pool,accuracy,parameters,seconds\n',results)
```
