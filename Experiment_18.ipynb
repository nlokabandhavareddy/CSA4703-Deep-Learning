```python
!pip install -q keras-tuner
import tensorflow as tf, numpy as np, matplotlib.pyplot as plt
from keras_tuner import RandomSearch
(x,y),(xt,yt)=tf.keras.datasets.imdb.load_data(num_words=5000); x=tf.keras.preprocessing.sequence.pad_sequences(x,maxlen=100); xt=tf.keras.preprocessing.sequence.pad_sequences(xt,maxlen=100)
def build(hp):
 typ=hp.Choice('cell',['RNN','LSTM','GRU']); units=hp.Int('units',16,64,step=16)
 m=tf.keras.Sequential([tf.keras.layers.Input((100,)),tf.keras.layers.Embedding(5000,32),{'RNN':tf.keras.layers.SimpleRNN,'LSTM':tf.keras.layers.LSTM,'GRU':tf.keras.layers.GRU}[typ](units),tf.keras.layers.Dropout(hp.Float('drop',0,.4,step=.2)),tf.keras.layers.Dense(1,activation='sigmoid')])
 m.compile('adam','binary_crossentropy',metrics=['accuracy']); return m
t=RandomSearch(build,objective='val_accuracy',max_trials=3,overwrite=True,directory='tuner',project_name='reviews'); t.search(x[:8000],y[:8000],epochs=2,validation_split=.2,verbose=0)
m=t.get_best_models(1)[0]; print('Best hyperparameters:',t.get_best_hyperparameters(1)[0].values); print('Test:',m.evaluate(xt[:2000],yt[:2000],verbose=0))
```
