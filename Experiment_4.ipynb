```python
import numpy as np, matplotlib.pyplot as plt
p=np.array([.5,.3,.2]); q1=np.array([.6,.25,.15]); q2=np.array([.2,.4,.4])
def entropy(a): return -(a*np.log2(a)).sum()
def ce(p,q): return -(p*np.log2(q)).sum()
def kl(p,q): return (p*np.log2(p/q)).sum()
H=entropy(p)
for q in [q1,q2]: print('H:',H,'CE:',ce(p,q),'KL:',kl(p,q),'H+KL:',H+kl(p,q))
print('KL asymmetric:',kl(p,q1),kl(q1,p))
x=np.linspace(.01,.99,100); y=-(.7*np.log(x)+.3*np.log(1-x))
plt.plot(x,y); plt.xlabel('Predicted probability'); plt.ylabel('Binary cross entropy'); plt.show()
```
