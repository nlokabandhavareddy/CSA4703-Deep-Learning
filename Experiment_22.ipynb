```python
import numpy as np, matplotlib.pyplot as plt
from scipy.signal import sawtooth
from sklearn.decomposition import FastICA,PCA
t=np.linspace(0,8,1000); S=np.c_[np.sin(2*t),np.sign(np.sin(3*t)),sawtooth(2*np.pi*t)]
A=np.random.default_rng(22).normal(size=(3,3)); M=S@A.T
ica=FastICA(n_components=3,random_state=1); R=ica.fit_transform(M); P=PCA(3).fit_transform(M)
fig,ax=plt.subplots(4,3,figsize=(12,7))
for j in range(3): ax[0,j].plot(S[:,j]); ax[1,j].plot(M[:,j]); ax[2,j].plot(R[:,j]); ax[3,j].plot(P[:,j])
for i,l in enumerate(['Original','Mixed','ICA','PCA']): ax[i,0].set_ylabel(l)
plt.tight_layout(); plt.show()
print('ICA source correlations (absolute):',np.abs(np.corrcoef(S.T,R.T)[:3,3:]))
```
