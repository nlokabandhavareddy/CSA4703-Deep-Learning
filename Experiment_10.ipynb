```python
import tensorflow as tf, numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
X,y=load_diabetes(return_X_y=True); y=(y>np.median(y)).astype(int); Xt,Xv,yt,yv=train_test_split(X,y,test_size=.25,random_state=3)
sc=StandardScaler(); Xt=sc.fit_transform(Xt); Xv=sc.transform(Xv)
results=[]
for label,reg,drop,es in [('Base',0,0,False),('L2',.01,0,False),('Dropout',0,.3,False),('EarlyStop',0,0,True)]:
 m=tf.keras.Sequential([tf.keras.layers.Input((10,)),tf.keras.layers.Dense(128,activation='relu',kernel_regularizer=tf.keras.regularizers.l2(reg)),tf.keras.layers.Dropout(drop),tf.keras.layers.Dense(64,activation='relu',kernel_regularizer=tf.keras.regularizers.l2(reg)),tf.keras.layers.Dense(1,activation='sigmoid')])
 m.compile('adam','binary_crossentropy',metrics=['accuracy']); callbacks=[tf.keras.callbacks.EarlyStopping(patience=3,restore_best_weights=True)] if es else []
 h=m.fit(Xt,yt,validation_split=.2,epochs=30,verbose=0,callbacks=callbacks)
 a=m.evaluate(Xv,yv,verbose=0)[1]; results.append((label,a,h.history['accuracy'][-1])); print(label,a)
print(results)
```
