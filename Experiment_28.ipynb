```python
import tensorflow as tf, numpy as np, matplotlib.pyplot as plt
# Small educational two-hidden-layer autoencoder/DBM-inspired representation
(x,y),(xt,yt)=tf.keras.datasets.mnist.load_data(); x=x[:5000].astype('float32')/255; x=x.reshape(-1,784)
m=tf.keras.Sequential([tf.keras.layers.Input((784,)),tf.keras.layers.Dense(128,activation='relu'),tf.keras.layers.Dense(64,activation='relu'),tf.keras.layers.Dense(128,activation='relu'),tf.keras.layers.Dense(784,activation='sigmoid')])
m.compile('adam','binary_crossentropy'); h=m.fit(x,x,epochs=3,batch_size=64,verbose=0)
rec=m.predict(x[:8],verbose=0); fig,ax=plt.subplots(2,8,figsize=(12,3))
for i in range(8): ax[0,i].imshow(x[i].reshape(28,28),cmap='gray'); ax[1,i].imshow(rec[i].reshape(28,28),cmap='gray'); ax[0,i].axis('off'); ax[1,i].axis('off')
plt.show(); print('Reconstruction loss:',h.history['loss'])
```
