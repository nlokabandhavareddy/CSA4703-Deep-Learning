```python
import numpy as np, matplotlib.pyplot as plt
from scipy.special import expit
rng=np.random.default_rng(26); v=64; h=16; W=rng.normal(0,.1,(v,h)); a=np.zeros(v); b=np.zeros(h)
vis=rng.random((10,v))
def gibbs(vv,k):
 for _ in range(k): hh=(rng.random((len(vv),h))<expit(vv@W+b)); vv=(rng.random((len(vv),v))<expit(hh@W.T+a))
 return vv
fig,ax=plt.subplots(1,4)
for i,k in enumerate([1,10,100,1000]): ax[i].imshow(gibbs(vis,k)[0].reshape(8,8)); ax[i].set_title(str(k)); ax[i].axis('off')
plt.show()
# Contrastive divergence k comparison toy
for k in [1,5,10]:
 chain=gibbs(vis,k); print('CD-k',k,'mean reconstruction mismatch',np.mean(abs(chain-vis)))
```
