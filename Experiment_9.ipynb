```python
import tensorflow as tf, numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report,confusion_matrix
X,y=load_breast_cancer(return_X_y=True); Xt,Xv,yt,yv=train_test_split(X,y,test_size=.2,stratify=y,random_state=2)
sc=StandardScaler(); Xt=sc.fit_transform(Xt); Xv=sc.transform(Xv)
m=tf.keras.Sequential([tf.keras.layers.Input((X.shape[1],)),tf.keras.layers.Dense(64,activation='relu'),tf.keras.layers.Dense(32,activation='relu'),tf.keras.layers.Dense(1,activation='sigmoid')])
m.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy']); h=m.fit(Xt,yt,validation_data=(Xv,yv),epochs=25,verbose=0)
p=(m.predict(Xv,verbose=0)>0.5).astype(int); print(classification_report(yv,p)); print(confusion_matrix(yv,p))
plt.plot(h.history['accuracy']); plt.plot(h.history['val_accuracy']); plt.legend(['Train','Val']); plt.show()
```
