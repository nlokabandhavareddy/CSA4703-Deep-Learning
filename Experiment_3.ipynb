```python
import numpy as np, matplotlib.pyplot as plt
from scipy.stats import bernoulli,binom,poisson,uniform,norm
rng=np.random.default_rng(3)
fig,ax=plt.subplots(1,5,figsize=(15,3))
samples=[bernoulli.rvs(.4,size=1000),binom.rvs(10,.4,size=1000),poisson.rvs(4,size=1000),uniform.rvs(size=1000),norm.rvs(size=1000)]
for a,s,t in zip(ax,samples,['Bernoulli','Binomial','Poisson','Uniform','Gaussian']): a.hist(s,bins=20); a.set_title(t)
plt.tight_layout(); plt.show()
prev=.01; sens=.95; spec=.90
post=sens*prev/(sens*prev+(1-spec)*(1-prev)); print('Bayes posterior:',post)
n=100000; disease=rng.random(n)<prev; positive=np.where(disease,rng.random(n)<sens,rng.random(n)<1-spec)
print('Simulated posterior:',disease[positive].mean())
p=np.linspace(.001,.2,100); postp=sens*p/(sens*p+(1-spec)*(1-p))
plt.plot(p,postp); plt.xlabel('Prevalence'); plt.ylabel('P(Disease|Positive)'); plt.show()
```
