```python
import tensorflow as tf, numpy as np, matplotlib.pyplot as plt, networkx as nx
m=tf.keras.Sequential([tf.keras.layers.Input((5,1)),tf.keras.layers.SimpleRNN(3,return_sequences=True),tf.keras.layers.Dense(1)]); m(np.zeros((1,5,1)))
Wx,Wh,b=m.layers[0].get_weights(); print('Input weights:',Wx,'\nRecurrent weights:',Wh,'\nBias:',b)
x=np.arange(1,6).reshape(1,5,1)/5; out=m(x).numpy().ravel(); print('Outputs:',out)
G=nx.DiGraph()
for t in range(5): G.add_node('x'+str(t+1)); G.add_node('h'+str(t+1)); G.add_edge('x'+str(t+1),'h'+str(t+1),label='Wxh');
for t in range(5):
 if t: G.add_edge('h'+str(t),'h'+str(t+1),label='Whh')
 G.add_edge('h'+str(t+1),'y'+str(t+1),label='Why')
pos={**{f'x{i+1}':(i,1) for i in range(5)},**{f'h{i+1}':(i,0) for i in range(5)},**{f'y{i+1}':(i,-1) for i in range(5)}}
nx.draw(G,pos,with_labels=True,node_size=900); plt.show()
plt.plot(out,'o-'); plt.title('Output across time'); plt.show()
```
