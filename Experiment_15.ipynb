```python
import tensorflow as tf, numpy as np, matplotlib.pyplot as plt
for L in [10,50,100]:
 x=tf.Variable(tf.random.normal((1,L,1)))
 r=tf.keras.layers.SimpleRNN(16,return_sequences=True); y=r(x); loss=tf.reduce_sum(y[:,-1,:])
 with tf.GradientTape() as tape: loss=tf.reduce_sum(r(x)[:,-1,:])
 g=tape.gradient(loss,x).numpy()[0,:,0]; plt.plot(np.abs(g),label=f'{L} steps')
plt.legend(); plt.xlabel('Time step'); plt.ylabel('Gradient magnitude'); plt.show()
print('Gradient magnitudes illustrate how gradients can diminish over long sequences.')
```
