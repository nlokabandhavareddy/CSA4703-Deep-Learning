```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import fetch_olivetti_faces
from sklearn.decomposition import PCA
X=fetch_olivetti_faces().data
mu=X.mean(0); Z=X-mu; cov=np.cov(Z,rowvar=False); w,v=np.linalg.eigh(cov); ix=np.argsort(w)[::-1]; w=w[ix]; v=v[:,ix]
pct=w/w.sum()*100; print('95% components:',np.searchsorted(np.cumsum(pct),95)+1)
p=PCA().fit(X); plt.plot(p.explained_variance_ratio_); plt.title('Scree plot'); plt.show(); plt.plot(np.cumsum(p.explained_variance_ratio_)); plt.axhline(.95); plt.show()
fig,ax=plt.subplots(1,5);
for i in range(5): ax[i].imshow(v[:,i].reshape(64,64),cmap='gray'); ax[i].axis('off')
plt.show()
for k in [10,50,100]:
 p=PCA(k).fit(X); rec=p.inverse_transform(p.transform(X[:1])); plt.imshow(rec.reshape(64,64),cmap='gray'); plt.title(f'{k} PCs'); plt.show()
```
