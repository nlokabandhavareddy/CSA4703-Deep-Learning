```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.decomposition import FactorAnalysis, PCA
rng=np.random.default_rng(21); latent=rng.normal(size=(300,3)); X=latent@rng.normal(size=(3,10))+rng.normal(size=(300,10)); df=pd.DataFrame(X,columns=[f'Q{i}' for i in range(1,11)])
print(df.corr())
for k in [2,3,4]:
 fa=FactorAnalysis(n_components=k).fit(X); L=fa.components_.T; print(k,'loadings:\n',L)
 plt.imshow(L,cmap='coolwarm'); plt.colorbar(); plt.title(f'{k} factors'); plt.show()
print('PCA variance:',PCA().fit(X).explained_variance_ratio_)
```
