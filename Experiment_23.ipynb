```python
!pip -q install pgmpy
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
import networkx as nx, matplotlib.pyplot as plt
model=DiscreteBayesianNetwork([('Smoking','Lung'),('Pollution','Lung'),('Lung','Cough'),('Lung','Xray'),('Lung','Fatigue')])
cpds=[TabularCPD('Smoking',2,[[.7],[.3]]),
TabularCPD('Pollution',2,[[.8],[.2]]),
TabularCPD('Lung',2,[[.95,.9,.8,.1],[.05,.1,.2,.9]],evidence=['Smoking','Pollution'],evidence_card=[2,2]),
TabularCPD('Cough',2,[[.8,.2],[.2,.8]],evidence=['Lung'],evidence_card=[2]),
TabularCPD('Xray',2,[[.9,.1],[.1,.9]],evidence=['Lung'],evidence_card=[2]),
TabularCPD('Fatigue',2,[[.7,.3],[.3,.7]],evidence=['Lung'],evidence_card=[2])]
model.add_cpds(*cpds); print('Valid:',model.check_model())

# Build a clean networkx DiGraph using only the edge relationships
gg = nx.DiGraph(list(model.edges()))
pos = nx.spring_layout(gg)
nx.draw(gg, pos=pos, with_labels=True, node_size=1800, arrows=True); plt.show()
```
